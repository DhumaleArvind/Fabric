```python
info = [("Phone","Apple",150000,"2019","14"),("TV","Sony",100000,"2024","BRAVIA 16")] 
# creating the column names
df1 = spark.createDataFrame(info,["Item","Brand","Price","Launchdate","Model"])
#info = [("Phone","Apple",150000,"2019","14"),("TV","Sony",100000,"2024","BRAVIA 16")] 
# printing the dataframe
display(df1)
```

```python
from pyspark.sql.functions import lit
from pyspark.sql.functions import when
df1=df1.withColumn("Unitssold",lit(None).cast("string"))
df1.show()
display(df1)
```

```python
df1 = df1.withColumn("Unitssold",when(df1.Item == "Phone",500).when(df1.Item == "TV",300).otherwise(0))
df1.show()
```

```python
display(df1)
```

```python
new_rows=[("Laptop","Dell",95000,"2025","G 16","400"),
           ("Phone","Oneplus",25000,"2026","ce6 lite","600") ]

df_new_rows = spark.createDataFrame(new_rows,df1.schema)
df1 = df1.unionByName(df_new_rows).dropDuplicates()
df1.show()
display(df1)
```

```python
df1.groupBy("Items").agg(min("Unitssold").alias("Min_of_Unitssold"))
```

```python
df1.groupBy("Item").agg(
    min(col("Unitssold")).alias("Min Units Sold"),
    max(col(("Unitssold")).alias("Max Units Sold"),
    avg(col("Unitssold")).alias("Avg Units Sold").show()
```
